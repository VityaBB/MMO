```python
# Задание 1: Найти максимальную цифру введенного натурального числа

def max_digit(number: int) -> int:
    """Возвращает максимальную цифру натурального числа."""
    return max(int(d) for d in str(number))

try:
    n = int(input("Введите натуральное число: "))
    if n <= 0:
        print("Число должно быть натуральным (положительным)!")
    else:
        print(f"Максимальная цифра числа {n}: {max_digit(n)}")
except ValueError:
    print("Ошибка: нужно ввести целое число!")
```

```python
# Задание 2: Удалить часть текста, заключённого в скобки

import re

def remove_brackets(text: str) -> str:
    """Удаляет все фрагменты текста, заключённые в круглые скобки."""
    return re.sub(r'\([^()]*\)', '', text)

user_text = input("Введите строку с текстом в скобках: ")
result = remove_brackets(user_text)

# Убираем лишние пробелы, оставшиеся после удаления
result = re.sub(r'\s+', ' ', result).strip()

print(f"\nИсходная строка: {user_text}")
print(f"Результат: {result}")
```

```python
# Задание 3: Найти наименьший чётный элемент; переставить нули в начало

def process_list(lst: list) -> tuple:
    """
    Возвращает:
      - наименьший чётный элемент (или первый элемент, если такого нет)
      - новый список: сначала нули, затем остальные
    """
    # Наименьший чётный элемент
    even_numbers = [x for x in lst if isinstance(x, int) and x % 2 == 0]
    min_even = min(even_numbers) if even_numbers else lst[0]

    # Перестановка: сначала нули, затем остальные
    zeros = [x for x in lst if x == 0]
    non_zeros = [x for x in lst if x != 0]
    new_list = zeros + non_zeros

    return min_even, new_list


my_list = [5, 0, 12, 3, 0, -8, 7, 0, 4, 9]
print(f"Исходный список: {my_list}")

min_even, reordered = process_list(my_list)
print(f"Наименьший чётный элемент: {min_even}")
print(f"Список после преобразования: {reordered}")
```

```python
# Задание 4: Словарь {символ: количество вхождений}

def char_frequency(text: str) -> dict:
    """Возвращает словарь: символ -> количество вхождений."""
    freq = {}
    for ch in text:
        freq[ch] = freq.get(ch, 0) + 1
    return freq


source_string = ' Follow your heart'
result_dict = char_frequency(source_string)

print(f"Исходная строка: '{source_string}'\n")
print("Словарь частот символов:")
for key, value in result_dict.items():
    print(f"  {repr(key)}: {value}")
```

```python
# Задание 5: Программа «Магазин игрушек»

# Словарь: название -> [состав, цена (руб), количество (шт)]
toys = {
    "Мяч":         ["Резина, диаметр 20 см", 350, 15],
    "Кукла":       ["Пластик, платье, 30 см", 1200, 8],
    "Конструктор": ["Пластик, 250 деталей", 2500, 5],
    "Машинка":     ["Металл, масштаб 1:24", 800, 12],
    "Пазл":        ["Картон, 1000 элементов", 600, 20],
    "Медведь":     ["Плюш, синтепон, 45 см", 1500, 6],
}


def show_description():
    print("\n--- Описание игрушек ---")
    for name, info in toys.items():
        print(f"{name}: {info[0]}")


def show_price():
    print("\n--- Цены ---")
    for name, info in toys.items():
        print(f"{name}: {info[1]} руб.")


def show_quantity():
    print("\n--- Количество на складе ---")
    for name, info in toys.items():
        print(f"{name}: {info[2]} шт.")


def show_all():
    print("\n--- Полная информация ---")
    for name, info in toys.items():
        print(f"{name}:\n  Состав: {info[0]}\n  Цена: {info[1]} руб.\n  Количество: {info[2]} шт.")


def buy():
    """Покупка товаров."""
    total_cost = 0
    purchase_list = []

    while True:
        name = input("\nВведите название игрушки (или 'n' для завершения покупки): ").strip()
        if name.lower() == 'n':
            break
        if name not in toys:
            print("Такой игрушки нет в магазине!")
            continue

        try:
            qty = int(input(f"Введите количество '{name}': "))
        except ValueError:
            print("Ошибка: нужно ввести целое число!")
            continue

        if qty <= 0:
            print("Количество должно быть больше нуля!")
            continue
        if qty > toys[name][2]:
            print(f"Недостаточно товара! В наличии: {toys[name][2]} шт.")
            continue

        price = toys[name][1]
        cost = price * qty
        total_cost += cost
        toys[name][2] -= qty
        purchase_list.append((name, qty, cost))
        print(f"Добавлено: {name} x {qty} = {cost} руб.")

    if purchase_list:
        print("\n--- Итог покупки ---")
        for name, qty, cost in purchase_list:
            print(f"  {name} x {qty} = {cost} руб.")
        print(f"Общая стоимость: {total_cost} руб.")
    else:
        print("Покупки не совершены.")

    print("\n--- Остатки на складе ---")
    for name, info in toys.items():
        print(f"  {name}: {info[2]} шт.")


def main():
    menu = {
        "1": ("Просмотр описания", show_description),
        "2": ("Просмотр цены", show_price),
        "3": ("Просмотр количества", show_quantity),
        "4": ("Вся информация", show_all),
        "5": ("Покупка", buy),
        "6": ("До свидания", None),
    }

    while True:
        print("\n========== МАГАЗИН ИГРУШЕК ==========")
        for key, (label, _) in menu.items():
            print(f"{key}. {label}")
        choice = input("Выберите пункт меню: ").strip()

        if choice not in menu:
            print("Неверный пункт меню! Попробуйте снова.")
            continue

        label, action = menu[choice]
        if choice == "6":
            print("До свидания! Ждём вас снова!")
            break
        action()


# Запуск программы
main()
```
