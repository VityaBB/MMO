```python
# Задание 1: Класс Human — имя, год рождения, 2 метода

from datetime import datetime


class Human:
    """Класс, описывающий человека."""

    def __init__(self, name: str, birth_year: int):
        """
        Конструктор класса.
        :param name: имя человека
        :param birth_year: год рождения
        """
        if not isinstance(name, str) or not name.strip():
            raise ValueError("Имя должно быть непустой строкой!")
        if not isinstance(birth_year, int):
            raise TypeError("Год рождения должен быть целым числом!")

        current_year = datetime.now().year
        if birth_year < 1900 or birth_year > current_year:
            raise ValueError(f"Год рождения должен быть в диапазоне 1900–{current_year}!")

        self.name = name
        self.birth_year = birth_year

    def show_info(self):
        """Метод 1: выводит имя и возраст."""
        age = datetime.now().year - self.birth_year
        print(f"Имя: {self.name}")
        print(f"Возраст: {age} лет")

    def is_adult(self) -> bool:
        """Метод 2: проверяет, является ли человек совершеннолетним (18+)."""
        age = datetime.now().year - self.birth_year
        return age >= 18


# --- Проверка работы класса ---
try:
    person1 = Human("Иван Петров", 2005)
    person2 = Human("Анна Сидорова", 2010)

    print("=== Человек 1 ===")
    person1.show_info()
    print(f"Совершеннолетний? {person1.is_adult()}")

    print("\n=== Человек 2 ===")
    person2.show_info()
    print(f"Совершеннолетний? {person2.is_adult()}")

    # Проверка исключительных ситуаций
    print("\n=== Проверка ошибок ===")
    try:
        bad = Human("", 2000)
    except ValueError as e:
        print(f"Ошибка: {e}")

    try:
        bad = Human("Тест", 1850)
    except ValueError as e:
        print(f"Ошибка: {e}")

except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
```

```python
# Задание 2: Класс Airline — рейсы, фильтрация по пункту и дню недели


class Airline:
    """Класс, описывающий авиарейс."""

    def __init__(self, destination: str, flight_number: str,
                 aircraft_type: str, departure_time: str, week_day: str):
        """
        :param destination: пункт назначения
        :param flight_number: номер рейса
        :param aircraft_type: тип самолёта
        :param departure_time: время вылета (например, "14:30")
        :param week_day: день недели (например, "Понедельник")
        """
        self.destination = destination
        self.flight_number = flight_number
        self.aircraft_type = aircraft_type
        self.departure_time = departure_time
        self.week_day = week_day

    def __str__(self):
        return (f"[{self.flight_number}] {self.destination} | "
                f"{self.aircraft_type} | {self.departure_time} | {self.week_day}")


def find_by_destination(flights: list, destination: str) -> list:
    """Возвращает список рейсов для заданного пункта назначения."""
    return [f for f in flights if f.destination.lower() == destination.lower()]


def find_by_weekday(flights: list, week_day: str) -> list:
    """Возвращает список рейсов для заданного дня недели."""
    return [f for f in flights if f.week_day.lower() == week_day.lower()]


# --- Создание списка рейсов ---
flights = [
    Airline("Москва",     "SU-101", "Boeing 737", "08:00", "Понедельник"),
    Airline("Париж",      "AF-225", "Airbus A320", "11:30", "Вторник"),
    Airline("Москва",     "SU-105", "Airbus A321", "15:45", "Среда"),
    Airline("Лондон",     "BA-879", "Boeing 777", "19:20", "Понедельник"),
    Airline("Москва",     "SU-109", "Boeing 737", "22:10", "Пятница"),
    Airline("Берлин",     "LH-310", "Airbus A319", "06:15", "Среда"),
    Airline("Париж",      "AF-230", "Boeing 787", "13:00", "Пятница"),
]

print("=== Все рейсы ===")
for f in flights:
    print(" ", f)

# --- Поиск по пункту назначения ---
print("\n=== Рейсы в Москву ===")
try:
    result = find_by_destination(flights, "Москва")
    if result:
        for f in result:
            print(" ", f)
    else:
        print("  Рейсов не найдено.")
except Exception as e:
    print(f"Ошибка: {e}")

# --- Поиск по дню недели ---
print("\n=== Рейсы в среду ===")
try:
    result = find_by_weekday(flights, "Среда")
    if result:
        for f in result:
            print(" ", f)
    else:
        print("  Рейсов не найдено.")
except Exception as e:
    print(f"Ошибка: {e}")
```

```python
# Задание 3: Иерархия классов Car -> TownCar, SportCar, WorkCar, PoliceCar


class Car:
    """Базовый класс автомобиля."""

    def __init__(self, speed: float, color: str, name: str, is_police: bool = False):
        """
        :param speed: текущая скорость (км/ч)
        :param color: цвет
        :param name: название модели
        :param is_police: является ли полицейской машиной
        """
        self.speed = speed
        self.color = color
        self.name = name
        self.is_police = is_police

    def go(self):
        print(f"{self.name}: машина поехала.")

    def stop(self):
        print(f"{self.name}: машина остановилась.")
        self.speed = 0

    def turn(self, direction: str):
        print(f"{self.name}: машина повернула {'направо' if direction.lower() == 'right' else 'налево'}.")

    def show_speed(self):
        """Базовый метод — просто показывает скорость."""
        print(f"{self.name}: текущая скорость = {self.speed} км/ч")


# ---------- Дочерние классы ----------

class TownCar(Car):
    """Городской автомобиль."""
    def show_speed(self):
        """Переопределён: предупреждение при превышении 60 км/ч."""
        print(f"{self.name} (TownCar): текущая скорость = {self.speed} км/ч")
        if self.speed > 60:
            print("  ⚠️ ВНИМАНИЕ: превышение скорости (лимит 60 км/ч)!")


class SportCar(Car):
    """Спортивный автомобиль."""
    pass


class WorkCar(Car):
    """Рабочий автомобиль."""
    def show_speed(self):
        """Переопределён: предупреждение при превышении 40 км/ч."""
        print(f"{self.name} (WorkCar): текущая скорость = {self.speed} км/ч")
        if self.speed > 40:
            print("  ⚠️ ВНИМАНИЕ: превышение скорости (лимит 40 км/ч)!")


class PoliceCar(Car):
    """Полицейский автомобиль."""
    def __init__(self, speed, color, name):
        # У полицейской машины всегда is_police = True
        super().__init__(speed, color, name, is_police=True)


# --- Демонстрация работы ---
cars = [
    TownCar(80, "белый", "Toyota Camry"),
    SportCar(220, "красный", "Ferrari F8"),
    WorkCar(55, "жёлтый", "ГАЗель"),
    PoliceCar(90, "синий", "Ford Mustang Police"),
]

print("=== Проверка атрибутов ===")
for car in cars:
    print(f"{car.name}: цвет={car.color}, скорость={car.speed}, полиция={car.is_police}")

print("\n=== Действия машин ===")
try:
    cars[0].go()
    cars[0].turn("right")
    cars[0].show_speed()
    cars[0].stop()

    print()
    cars[2].go()
    cars[2].show_speed()
    cars[2].stop()

    print()
    cars[3].go()
    cars[3].show_speed()
    cars[3].turn("left")
except Exception as e:
    print(f"Ошибка: {e}")
```

```python
# Задание 4: Класс, демонстрирующий методы экземпляра, статические и методы класса
# Университет: БГУИР


class Student:
    """Класс, описывающий студента БГУИР."""

    # ---------- АТРИБУТЫ КЛАССА (общие для всех объектов) ----------
    total_students = 0                  # счётчик всех созданных студентов
    university = "БГУИР"                # название университета

    def __init__(self, name: str, group: str, gpa: float):
        """
        Конструктор (метод экземпляра).
        :param name:  ФИО студента
        :param group: номер группы
        :param gpa:   средний балл (0–10)
        """
        if not isinstance(name, str) or not name.strip():
            raise ValueError("Имя должно быть непустой строкой!")
        if not self.is_valid_gpa(gpa):
            raise ValueError("Средний балл должен быть числом от 0 до 10!")

        self.name = name
        self.group = group
        self.gpa = gpa
        Student.total_students += 1     # увеличиваем общий счётчик

    # ---------- МЕТОДЫ ЭКЗЕМПЛЯРА ----------
    def show_info(self):
        """Метод экземпляра: выводит информацию о студенте БГУИР."""
        print(f"Студент: {self.name}")
        print(f"  Университет: {Student.university}")
        print(f"  Группа: {self.group}")
        print(f"  Средний балл: {self.gpa}")

    def is_excellent(self) -> bool:
        """Метод экземпляра: отличник ли студент (средний балл ≥ 9.0)."""
        return self.gpa >= 9.0

    # ---------- СТАТИЧЕСКИЙ МЕТОД ----------
    @staticmethod
    def is_valid_gpa(gpa) -> bool:
        """
        Статический метод: проверяет корректность среднего балла.
        Не зависит ни от класса, ни от конкретного экземпляра.
        """
        return isinstance(gpa, (int, float)) and 0 <= gpa <= 10

    # ---------- МЕТОДЫ КЛАССА ----------
    @classmethod
    def change_university(cls, new_name: str):
        """Метод класса: меняет название университета для всех студентов."""
        cls.university = new_name
        print(f"Университет изменён на: {new_name}")

    @classmethod
    def get_total(cls) -> int:
        """Метод класса: возвращает общее количество созданных студентов."""
        return cls.total_students

    @classmethod
    def reset_counter(cls):
        """Метод класса: сбрасывает счётчик студентов (для повторных тестов)."""
        cls.total_students = 0


# --- Демонстрация работы ---
print("=== Создание студентов БГУИР ===")
try:
    s1 = Student("Иванов И.И.", "ПО-21", 9.5)
    s2 = Student("Петров П.П.", "ПО-21", 7.2)
    s3 = Student("Сидорова А.А.", "ПО-22", 8.9)
    print(f"Всего студентов: {Student.get_total()}")
except ValueError as e:
    print(f"Ошибка при создании: {e}")
finally:
    print("--- Блок создания завершён ---")

print("\n=== Информация о студентах (метод экземпляра) ===")
s1.show_info()
print(f"Отличник? {s1.is_excellent()}")

print()
s2.show_info()
print(f"Отличник? {s2.is_excellent()}")

print("\n=== Статический метод ===")
for gpa in [9.5, -1, 11, "abc"]:
    print(f"  Балл {gpa!r}: корректен? {Student.is_valid_gpa(gpa)}")

print("\n=== Метод класса: смена университета ===")
Student.change_university("БГУИР")   # подтверждаем название
s1.show_info()                       # покажет БГУИР

print(f"\nВсего студентов (метод класса): {Student.get_total()}")

# --- Проверка исключительных ситуаций ---
print("\n=== Проверка ошибок ===")
try:
    bad = Student("", "ПО-21", 8.0)
except ValueError as e:
    print(f"Ошибка: {e}")

try:
    bad = Student("Тестов Т.Т.", "ПО-21", 15)
except ValueError as e:
    print(f"Ошибка: {e}")
```
