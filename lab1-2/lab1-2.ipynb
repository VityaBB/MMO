```python
# Задание 1: Функция bank(a, years) — расчёт суммы вклада под 10% годовых

def bank(a, years):
    """
    Принимает:
        a     — начальная сумма вклада (руб.)
        years — срок вклада (лет)
    Возвращает:
        итоговую сумму на счёте (сложный процент 10% годовых)
    """
    if not isinstance(a, (int, float)) or not isinstance(years, int):
        raise TypeError("Сумма должна быть числом, а срок — целым числом лет!")
    if a < 0:
        raise ValueError("Сумма вклада не может быть отрицательной!")
    if years < 0:
        raise ValueError("Срок вклада не может быть отрицательным!")

    total = a
    for _ in range(years):
        total *= 1.10          # +10% каждый год
    return round(total, 2)


# --- Проверка работы функции с обработкой исключений ---
try:
    a = float(input("Введите сумму вклада (руб.): "))
    years = int(input("Введите срок вклада (лет): "))
    result = bank(a, years)
    print(f"\nЧерез {years} лет на счёте будет: {result} руб.")
except ValueError as e:
    print(f"Ошибка ввода: {e}")
except TypeError as e:
    print(f"Ошибка типа данных: {e}")
except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
finally:
    print("--- Расчёт завершён ---")
```

```python
# Задание 2: Универсальная функция для set / list / number / string

def universal(arg):
    """
    Обрабатывает аргумент в зависимости от его типа:
      - set    → сумма всех элементов
      - list   → произведение между первым и вторым отрицательными;
                 макс и мин элементы поменять местами
      - число  → сумма цифр
      - строка → самое длинное слово
    """
    # --------- МНОЖЕСТВО ---------
    if isinstance(arg, set):
        if not arg:
            return "Множество пустое"
        return f"Сумма всех элементов множества: {sum(arg)}"

    # --------- СПИСОК ---------
    elif isinstance(arg, list):
        if not arg:
            return "Список пуст"

        result = []

        # 1) Произведение между первым и вторым отрицательными
        neg_indices = [i for i, x in enumerate(arg) if isinstance(x, (int, float)) and x < 0]
        if len(neg_indices) >= 2:
            i1, i2 = neg_indices[0], neg_indices[1]
            between = arg[i1 + 1:i2]
            if between:
                product = 1
                for x in between:
                    if isinstance(x, (int, float)):
                        product *= x
                result.append(f"Произведение между 1-м и 2-м отриц.: {product}")
            else:
                result.append("Между отрицательными нет элементов")
        else:
            result.append("Меньше двух отрицательных элементов")

        # 2) Поменять местами max и min
        numbers = [x for x in arg if isinstance(x, (int, float))]
        if numbers:
            new_list = arg.copy()
            idx_max = new_list.index(max(numbers))
            idx_min = new_list.index(min(numbers))
            new_list[idx_max], new_list[idx_min] = new_list[idx_min], new_list[idx_max]
            result.append(f"Список после обмена max↔min: {new_list}")

        return "\n".join(result)

    # --------- ЧИСЛО ---------
    elif isinstance(arg, int) and not isinstance(arg, bool):
        digits_sum = sum(int(d) for d in str(abs(arg)))
        return f"Сумма цифр числа {arg}: {digits_sum}"

    # --------- СТРОКА ---------
    elif isinstance(arg, str):
        words = arg.split()
        if not words:
            return "Строка пуста"
        longest = max(words, key=len)
        return f"Самое длинное слово: '{longest}' (длина {len(longest)})"

    else:
        return f"Тип {type(arg).__name__} не поддерживается"


# --- Проверка всех случаев с try/except/finally ---
test_cases = [
    {1, 2, 3, 4, 5},
    [5, -1, 3, -2, 8, 10, -7],
    12345,
    "Python — лучший язык программирования",
]

for case in test_cases:
    print(f"\n>>> Аргумент: {case!r}  (тип: {type(case).__name__})")
    try:
        print(universal(case))
    except Exception as e:
        print(f"Ошибка: {e}")
    finally:
        print("-" * 40)
```

```python
# Задание 3: Найти сумму каждого столбца матрицы m×n

def column_sums(matrix):
    """Возвращает список сумм элементов по каждому столбцу."""
    if not matrix:
        raise ValueError("Матрица пуста!")
    if not all(isinstance(row, list) for row in matrix):
        raise TypeError("Матрица должна быть списком списков!")
    if len({len(row) for row in matrix}) != 1:
        raise ValueError("Все строки матрицы должны быть одинаковой длины!")

    n_cols = len(matrix[0])
    sums = [0] * n_cols
    for row in matrix:
        for j, val in enumerate(row):
            sums[j] += val
    return sums


def print_matrix(matrix):
    for row in matrix:
        print("  ", row)


# --- Ввод матрицы с обработкой исключений ---
try:
    m = int(input("Введите количество строк m: "))
    n = int(input("Введите количество столбцов n: "))

    if m <= 0 or n <= 0:
        raise ValueError("Размеры матрицы должны быть положительными!")

    matrix = []
    print(f"\nВведите {m} строк по {n} чисел (через пробел):")
    for i in range(m):
        while True:
            try:
                row = list(map(int, input(f"Строка {i + 1}: ").split()))
                if len(row) != n:
                    raise ValueError(f"Нужно ввести ровно {n} чисел!")
                matrix.append(row)
                break
            except ValueError as e:
                print(f"  Ошибка: {e}. Попробуйте снова.")

    print("\nИсходная матрица:")
    print_matrix(matrix)

    sums = column_sums(matrix)
    print("\nСуммы по столбцам:")
    for j, s in enumerate(sums, start=1):
        print(f"  Столбец {j}: {s}")

except ValueError as e:
    print(f"Ошибка: {e}")
except TypeError as e:
    print(f"Ошибка типа: {e}")
except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
finally:
    print("\n--- Программа завершена ---")
```

```python
# Задание 4: Программа, демонстрирующая работу try / except / finally

def divide(a, b):
    """Деление с проверкой исключительных ситуаций."""
    print(f"  [внутри функции] a={a}, b={b}")
    try:
        result = a / b
    except ZeroDivisionError:
        print("  ❌ Перехвачено: деление на ноль!")
        return None
    except TypeError:
        print("  ❌ Перехвачено: неверный тип данных!")
        return None
    else:
        print("  ✅ Деление прошло успешно.")
        return result
    finally:
        print("  🔹 Блок finally выполнен (всегда).")


def demo_try_except_finally():
    """Полная демонстрация работы try/except/finally."""
    print("=" * 50)
    print("ДЕМОНСТРАЦИЯ try / except / finally")
    print("=" * 50)

    # Случай 1: корректные данные
    print("\n--- Случай 1: корректные данные ---")
    try:
        a = float(input("Введите делимое: "))
        b = float(input("Введите делитель: "))
        result = divide(a, b)
        if result is not None:
            print(f"  Результат: {a} / {b} = {result}")
    except ValueError:
        print("  ❌ Введено не число!")
    finally:
        print("  🔹 Внешний finally: блок 1 завершён.")

    # Случай 2: деление на ноль
    print("\n--- Случай 2: деление на ноль ---")
    try:
        result = divide(10, 0)
    finally:
        print("  🔹 Внешний finally: блок 2 завершён.")

    # Случай 3: неверный тип
    print("\n--- Случай 3: неверный тип данных ---")
    try:
        result = divide("10", 2)
    finally:
        print("  🔹 Внешний finally: блок 3 завершён.")

    # Случай 4: обращение к несуществующему индексу
    print("\n--- Случай 4: IndexError ---")
    try:
        lst = [1, 2, 3]
        print(lst[10])
    except IndexError as e:
        print(f"  ❌ Перехвачено: {e}")
    finally:
        print("  🔹 Внешний finally: блок 4 завершён.")


# Запуск демонстрации
demo_try_except_finally()
```
